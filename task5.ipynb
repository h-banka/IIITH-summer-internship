```python
model = YOLO("runs/detect/crowd_density_model/weights/best.pt")

results = model.predict(
    source="crowdhuman-1/test/images",
    conf=0.25,
    save=True
)
```

```python
ls runs/detect
```

```python
for r in results:
    print("People detected:", len(r.boxes))
```

```python
def density_label(count):
    if count < 10:
        return "LOW"
    elif count < 30:
        return "MEDIUM"
    else:
        return "HIGH"

for r in results:
    count = len(r.boxes)
    print("Density:", density_label(count))
```
